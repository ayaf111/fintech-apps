```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

df = pd.read_csv("with_clean_text.csv")
df.dropna(subset=['clean_text'], inplace=True)
```

```python
df = pd.read_csv("with_transalted_review.csv")

def assign_theme(text):
    themes = []
    text = str(text).lower()

    if any(word in text for word in ['login', 'access', 'password', 'authentication', 'sign', 'log']):
        themes.append('Account Access')

    if any(word in text for word in ['crash', 'slow', 'bug', 'freeze', 'lag', 'hang', 'glitch', 'fix', 'update', 'version', 'network', 'system']):
        themes.append('UI/UX')

    if any(word in text for word in ['transfer', 'payment', 'delay', 'deposit', 'withdraw', 'transaction', 'send', 'money']):
        themes.append('Transaction')

    if any(word in text for word in ['support', 'help', 'rude', 'response', 'agent', 'customer', 'service']):
        themes.append('Customer Support')

    if any(word in text for word in ['safe', 'security', 'fraud', 'unauthorized', 'scam', 'secure']):
        themes.append('Security')

    if any(word in text for word in ['feature', 'add', 'missing', 'option', 'wish', 'would like', 'need']):
        themes.append('Feature Request')

    if any(word in text for word in [
        'good', 'nice', 'amazing', 'great', 'excellent', 'super', 'wow', 'love',
        'thank', 'fantastic', 'perfect', 'awesome', 'best']):
        themes.append('Positive Feedback')

    if any(word in text for word in [
        'bad', 'poor', 'worst', 'disappoint', 'problem', 'useless']):
        themes.append('Negative Feedback')

    if any(word in text for word in [
        'easy', 'use', 'friendly', 'convenient', 'simple', 'functional', 'smart']):
        themes.append('User Experience')

    return ', '.join(themes) if themes else 'Other'

df['theme'] = df['clean_text'].apply(assign_theme)
```

```python
from googletrans import Translator

translator = Translator()
```

```python
def translate_to_english(text):
    try:
        if pd.isna(text):
            return text
        translated = translator.translate(str(text), src='auto', dest='en')
        return translated.text
    except Exception as e:
        return text  # fallback to original if translation fails
```

```python
# Filter only the 'Other' themed reviews
df_other = df[df['theme'] == 'Other'].copy()

# Translate and store in a new column
df_other['translated_review'] = df_other['review_text'].apply(translate_to_english)
```
